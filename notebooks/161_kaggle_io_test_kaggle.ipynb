```python
# Cell 1: write basic info to file
import os, sys, time, subprocess
TRACE = open('/kaggle/working/trace.log', 'w')
def T(msg):
    TRACE.write(f'[{time.strftime("%H:%M:%S")}] {msg}\n'); TRACE.flush()
    print(msg, flush=True)
T('CELL_1_START')
T(f'python={sys.version}')
T(f'cwd={os.getcwd()}')
T(f'listdir_working={os.listdir("/kaggle/working")}')
T('CELL_1_END')
TRACE.flush()
```
