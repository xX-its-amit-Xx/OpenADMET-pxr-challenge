```python
import os, time, sys
print('=== HELLO FROM CELL 1 ===', flush=True)
with open('/kaggle/working/hello.txt', 'w') as f:
    f.write(f'hello {time.time()}\n')
    f.write(f'python={sys.version}\n')
    f.write(f'cwd={os.getcwd()}\n')
    f.write(f'listdir={os.listdir("/kaggle/working")}\n')
print('wrote hello.txt', flush=True)
```

```python
print('=== CELL 2 GPU TEST ===', flush=True)
with open('/kaggle/working/gpu.txt', 'w') as f:
    try:
        import torch
        f.write(f'torch={torch.__version__}\n')
        f.write(f'cuda_available={torch.cuda.is_available()}\n')
        if torch.cuda.is_available():
            f.write(f'device={torch.cuda.get_device_name(0)}\n')
    except Exception as e:
        f.write(f'ERR: {type(e).__name__}: {e}\n')
print('wrote gpu.txt', flush=True)
```
